## Analyze 'in degree' and 'out degree'

### Subtask:
For each key (50, 100, 200), analyze the 'in_degree' and 'out_degree' distributions. This might involve calculating summary statistics (mean, median, standard deviation) and potentially visualizing the distributions (histograms or box plots).


**Reasoning**:
I need to iterate through the keys of the dictionary, access the 'in_degree' and 'out_degree' data, calculate and print summary statistics, and plot histograms for each distribution.



```python
import matplotlib.pyplot as plt
import pandas as pd # Import pandas to create Series

for key in df.keys():
    print(f"Analysis for key: {key}")

    # Access the actual data from the 'values' key and convert to pandas Series
    in_degree_data = pd.Series(df[key]['in_degree']['values'])
    print("In-Degree Statistics:")
    print(f"  Mean: {in_degree_data.mean():.2f}")
    print(f"  Median: {in_degree_data.median():.2f}")
    print(f"  Standard Deviation: {in_degree_data.std():.2f}")

    # Access the actual data from the 'values' key and convert to pandas Series
    out_degree_data = pd.Series(df[key]['out_degree']['values'])
    print("Out-Degree Statistics:")
    print(f"  Mean: {out_degree_data.mean():.2f}")
    print(f"  Median: {out_degree_data.median():.2f}")
    print(f"  Standard Deviation: {out_degree_data.std():.2f}")

    # Visualize in_degree distribution
    plt.figure(figsize=(10, 4))
    plt.subplot(1, 2, 1)
    in_degree_data.hist(bins=50)
    plt.title(f'{key} - In-Degree Distribution')
    plt.xlabel('Degree')
    plt.ylabel('Frequency')


    # Visualize out_degree distribution
    plt.subplot(1, 2, 2)
    out_degree_data.hist(bins=50)
    plt.title(f'{key} - Out-Degree Distribution')
    plt.xlabel('Degree')
    plt.ylabel('Frequency')

    plt.tight_layout()
    plt.show()

    print("-" * 30)
```

## Analyze 'wcc' and 'scc'

### Subtask:
For each key (50, 100, 200), analyze the properties of the weakly and strongly connected components. This could involve determining the number and size of components, and potentially visualizing the size distribution.


**Reasoning**:
Iterate through the keys, access the 'wcc' and 'scc' data, calculate the size of each component, and print summary statistics and visualize the size distribution with histograms.



```python
import matplotlib.pyplot as plt
import numpy as np

for key in df.keys():
    print(f"Analysis for key: {key}")

    # Analyze WCC
    wcc_data = df[key]['wcc']
    wcc_component_sizes = [len(component) for component in wcc_data]
    wcc_component_sizes = pd.Series(wcc_component_sizes)

    print("Weakly Connected Components (WCC) Statistics:")
    print(f"  Number of components: {len(wcc_component_sizes)}")
    print(f"  Mean size: {wcc_component_sizes.mean():.2f}")
    print(f"  Median size: {wcc_component_sizes.median():.2f}")
    print(f"  Standard Deviation of size: {wcc_component_sizes.std():.2f}")
    print(f"  Min size: {wcc_component_sizes.min()}")
    print(f"  Max size: {wcc_component_sizes.max()}")


    # Analyze SCC
    scc_data = df[key]['scc']
    scc_component_sizes = [len(component) for component in scc_data]
    scc_component_sizes = pd.Series(scc_component_sizes)

    print("\nStrongly Connected Components (SCC) Statistics:")
    print(f"  Number of components: {len(scc_component_sizes)}")
    print(f"  Mean size: {scc_component_sizes.mean():.2f}")
    print(f"  Median size: {scc_component_sizes.median():.2f}")
    print(f"  Standard Deviation of size: {scc_component_sizes.std():.2f}")
    print(f"  Min size: {scc_component_sizes.min()}")
    print(f"  Max size: {scc_component_sizes.max()}")


    # Visualize WCC and SCC size distributions
    plt.figure(figsize=(12, 5))

    plt.subplot(1, 2, 1)
    wcc_component_sizes.hist(bins=20)
    plt.title(f'{key} - WCC Component Size Distribution')
    plt.xlabel('Component Size')
    plt.ylabel('Frequency')

    plt.subplot(1, 2, 2)
    scc_component_sizes.hist(bins=20)
    plt.title(f'{key} - SCC Component Size Distribution')
    plt.xlabel('Component Size')
    plt.ylabel('Frequency')

    plt.tight_layout()
    plt.show()

    print("-" * 30)
```

## Analyze 'hop plot' and 'hop plot wcc'

### Subtask:
For each key (50, 100, 200), analyze the hop plot data. This involves understanding how reachability changes with distance and visualizing the hop plots. This is a retry due to previous failure in extracting hop distance data.


**Reasoning**:
Iterate through each key, access the hop plot and hop plot WCC data, create hop distance lists, and plot the reachability against hop distance for each. Also, print the summary statistics for the reachability values.



```python
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np

for key in df.keys():
    print(f"Analysis for key: {key}")

    # Access the hop plot data
    hop_plot_data = df[key]['hop_plot']
    hop_plot_values = pd.Series(hop_plot_data['values'])
    hop_distances = pd.Series(range(len(hop_plot_values)))

    # Access the hop plot WCC data
    hop_plot_wcc_data = df[key]['hop_plot_wcc']
    hop_plot_wcc_values = pd.Series(hop_plot_wcc_data['values'])
    hop_distances_wcc = pd.Series(range(len(hop_plot_wcc_values)))

    # Plot hop plot
    plt.figure(figsize=(12, 5))
    plt.subplot(1, 2, 1)
    plt.plot(hop_distances, hop_plot_values)
    plt.title(f'{key} - Hop Plot')
    plt.xlabel('Hop Distance')
    plt.ylabel('Reachability')
    plt.grid(True)

    # Plot hop plot WCC
    plt.subplot(1, 2, 2)
    plt.plot(hop_distances_wcc, hop_plot_wcc_values)
    plt.title(f'{key} - Hop Plot WCC')
    plt.xlabel('Hop Distance')
    plt.ylabel('Reachability')
    plt.grid(True)

    plt.tight_layout()
    plt.show()

    # Print summary statistics
    print("Hop Plot Statistics:")
    print(f"  Mean: {hop_plot_values.mean():.2f}")
    print(f"  Median: {hop_plot_values.median():.2f}")
    print(f"  Min: {hop_plot_values.min():.2f}")
    print(f"  Max: {hop_plot_values.max():.2f}")
    print(f"  Standard Deviation: {hop_plot_values.std():.2f}")

    print("\nHop Plot WCC Statistics:")
    print(f"  Mean: {hop_plot_wcc_values.mean():.2f}")
    print(f"  Median: {hop_plot_wcc_values.median():.2f}")
    print(f"  Min: {hop_plot_wcc_values.min():.2f}")
    print(f"  Max: {hop_plot_wcc_values.max():.2f}")
    print(f"  Standard Deviation: {hop_plot_wcc_values.std():.2f}")

    print("-" * 30)
```

## Analyze 'singular vector' and 'singular values'

### Subtask:
For each key (50, 100, 200), analyze the singular vector and singular values. This might involve examining the distribution of values and their significance.


**Reasoning**:
Iterate through the keys, access the singular values and vectors, calculate and print summary statistics for singular values, visualize singular values, and print the first few singular vector elements.



```python
import matplotlib.pyplot as plt
import numpy as np

for key in df.keys():
    print(f"Analysis for key: {key}")

    # Access singular values and vector
    singular_values = pd.Series(df[key]['singular_values']['values'])
    singular_vector = df[key]['singular_vector']['values']

    # Analyze singular values
    print("\nSingular Values Statistics:")
    print(f"  Mean: {singular_values.mean():.4f}")
    print(f"  Median: {singular_values.median():.4f}")
    print(f"  Standard Deviation: {singular_values.std():.4f}")
    print(f"  Min: {singular_values.min():.4f}")
    print(f"  Max: {singular_values.max():.4f}")

    # Visualize singular values
    plt.figure(figsize=(10, 4))
    plt.subplot(1, 2, 1)
    singular_values.hist(bins=50)
    plt.title(f'{key} - Singular Values Distribution')
    plt.xlabel('Singular Value')
    plt.ylabel('Frequency')

    plt.subplot(1, 2, 2)
    plt.plot(sorted(singular_values, reverse=True))
    plt.title(f'{key} - Singular Values (Descending Order)')
    plt.xlabel('Index')
    plt.ylabel('Singular Value')

    plt.tight_layout()
    plt.show()

    # Examine singular vector
    print("\nFirst 5 elements of Singular Vector:")
    display(singular_vector[:5])

    print("-" * 30)
```
